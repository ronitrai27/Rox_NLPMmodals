```python
import torch
import transformers
from transformers import AutoTokenizer, AutoModelForQuestionAnswering, pipeline
```

```python
qa_input = [{'question' : 'Why is coding important?' ,
             'context' : 'The coding is not boring inspite i love to code and to dance and coding important to find solution to problems.'},
            {'question' : 'How many programming languages does BLOOM support?',
             'context' : 'BLOOM has 176 billion parameters and can generate 48 programming languages as of now.'},
            {'question' : 'Who is the most handsome in the room?',
             'context' : 'The ROOM was filled with girls and lot of boys and many beautiful flowers but the most handsome was Ronit rai.'},
            {'question': 'What is the significance of data science in today’s world?',
     'context': 'In the current era of digitalization,  Data science plays a crucial role in making informed decisions. With an ever-increasing amount of data generated daily, data science helps in extracting valuable insights from raw information.'}]
```

```python
model_name = "deepset/roberta-base-squad2"
```

```python
model = AutoModelForQuestionAnswering.from_pretrained(model_name)
tokenizer = AutoTokenizer.from_pretrained(model_name)
```

```python
inputs0 = tokenizer(qa_input[0]['question'], qa_input[0]['context'], return_tensors='pt')
output0 = model(**inputs0)
inputs1 = tokenizer(qa_input[1]['question'], qa_input[1]['context'], return_tensors='pt')
output1 = model(**inputs1)
inputs2 = tokenizer(qa_input[2]['question'], qa_input[2]['context'], return_tensors='pt')
output2 = model(**inputs2)
inputs3 = tokenizer(qa_input[3]['question'], qa_input[3]['context'], return_tensors='pt')
output3 = model(**inputs3)

```

```python
answer_start_index = torch.argmax(output0.start_logits)
answer_end_index = torch.argmax(output0.end_logits)

answer_tokens = inputs0.input_ids[0, answer_start_index : answer_end_index + 1]
answer = tokenizer.decode(answer_tokens)
print("ques: {}\nanswer : {}".format(qa_input[0]['question'], answer))
```

```python
answer_start_index = torch.argmax(output1.start_logits)
answer_end_index = torch.argmax(output1.end_logits)

answer_tokens = inputs1.input_ids[0, answer_start_index : answer_end_index + 1]
answer = tokenizer.decode(answer_tokens)
print("ques: {}\nanswer : {}".format(qa_input[1]['question'], answer))
```

```python
answer_start_index = torch.argmax(output2.start_logits)
answer_end_index = torch.argmax(output2.end_logits)

answer_tokens = inputs2.input_ids[0, answer_start_index : answer_end_index + 1]
answer = tokenizer.decode(answer_tokens)
print("ques: {}\nanswer : {}".format(qa_input[2]['question'], answer))
```

```python
answer_start_index = torch.argmax(output3.start_logits)
answer_end_index = torch.argmax(output3.end_logits)

answer_tokens = inputs3.input_ids[0, answer_start_index : answer_end_index + 1]
answer = tokenizer.decode(answer_tokens)
print("ques: {}\nanswer : {}".format(qa_input[3]['question'], answer))
```
